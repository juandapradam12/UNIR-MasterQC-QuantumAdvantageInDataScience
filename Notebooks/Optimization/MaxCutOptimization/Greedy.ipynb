# MaxCut — Greedy (local search)

Heurística: partir de una asignación aleatoria fija (seed) y **voltear** el vértice
que más aumenta el corte, hasta un óptimo local.

Misma instancia HamLib $n=6$ que el resto de solvers MaxCut.


```python
import itertools
import json
import math
import random
import time
from pathlib import Path

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
```

```python
def resolve_maxcut_dir() -> Path:
    for base in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
        root = base / "Notebooks" / "Optimization" / "MaxCutOptimization"
        if root.is_dir():
            return root
    return Path.cwd().resolve() / "Notebooks" / "Optimization" / "MaxCutOptimization"


MAXCUT_DIR = resolve_maxcut_dir()
INSTANCE_PATH = MAXCUT_DIR / "instance_hamlib.json"
METRICS_DIR = MAXCUT_DIR / "metrics"
METRICS_DIR.mkdir(parents=True, exist_ok=True)

instance = json.loads(INSTANCE_PATH.read_text())
n = int(instance["n"])
edges = [tuple(e) for e in instance["edges"]]
OPTIMAL_CUT = int(instance["optimal_cut"])
OPTIMAL_BITS = list(instance["optimal_bits"])
nodes = list(range(n))

print(
    f"HamLib MaxCut | index={instance['hamlib_index']} id={instance['hamlib_id']} "
    f"n={n} |E|={len(edges)} optimal_cut={OPTIMAL_CUT}"
)
print("Fuente:", instance["url"])
print("Demo QAOA:", instance["demo_ref"])


def cut_value(bits) -> int:
    return int(sum(1 for a, b in edges if bits[a] != bits[b]))


def is_success(bits) -> bool:
    return cut_value(bits) == OPTIMAL_CUT


def save_maxcut_metrics(payload: dict, filename: str):
    path = METRICS_DIR / filename
    path.write_text(json.dumps(payload, indent=2))
    print("Saved:", path)
    return path
```

## Resolución


```python
def greedy_maxcut(seed=0):
    random.seed(seed)
    bits = [random.randint(0, 1) for _ in range(n)]
    steps = 0
    improved = True
    while improved:
        improved = False
        best_delta, best_v = 0, None
        base = cut_value(bits)
        for v in range(n):
            bits[v] ^= 1
            delta = cut_value(bits) - base
            bits[v] ^= 1
            if delta > best_delta:
                best_delta, best_v = delta, v
        if best_v is not None and best_delta > 0:
            bits[best_v] ^= 1
            steps += 1
            improved = True
    return bits, steps

t0 = time.perf_counter()
bits, steps = greedy_maxcut(seed=0)
elapsed = time.perf_counter() - t0
c = cut_value(bits)
success = is_success(bits)
print("Solución:", bits, "cut =", c, "factible/éxito =", success)

payload = {
    "problem": "maxcut",
    "solver": "greedy",
    "family": "classical",
    "protocol": {
        "n_runs": 1,
        "method": "local_search_best_flip",
        "sense": "maximize_cut",
        "hamlib_index": instance["hamlib_index"],
        "hamlib_id": instance["hamlib_id"],
    },
    "instance": {"n": n, "edges": [list(e) for e in edges], "source": instance["source"]},
    "quality": {
        "optimal_cost": float(OPTIMAL_CUT),
        "optimal_bits": OPTIMAL_BITS,
        "cost_mean": float(c),
        "cost_best": float(c),
        "gap_mean": float(OPTIMAL_CUT - c),
    },
    "success_rate": float(success),
    "time_s": {"mean": float(elapsed), "std": 0.0, "total": float(elapsed)},
    "effort": {"flip_steps": steps},
    "runs": [{"seed": 0, "bits": bits, "cut": float(c), "success": bool(success), "time_s": float(elapsed)}],
}
save_maxcut_metrics(payload, "greedy_maxcut.json")
```
